```python
import sqlite3
```

```python
conn = sqlite3.connect('business_analysis.db')
```

```python
df.to_sql('sales_data', conn, if_exists='replace', index=False)
```

```python
query = """
SELECT *
FROM sales_data
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT Product,
       SUM(Total_Sales) AS Total_Sales
       FROM sales_data
       GROUP BY Product
       ORDER BY Total_Sales DESC
       LIMIT 5;
       """

result = pd.read_sql_query(query, conn)

result
```
